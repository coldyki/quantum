# Eigenstate

## goal

In this notebook, i explore eignestates and eigenvalues of quantum gate using Qiskit

The main question is:

what happen when a quantum gate acts on a eigenstate

For an eigenstate  |$\psi$> of an operator U

U|$\psi$> = $\lambda$|$\psi$>

the eigenvalue $\lambda$ can contain a phase : 
$\lambda$ = e^i $\phi$



```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)
qc.z(0)

state = Statevector.from_instruction(qc)

print(state)
```

```python
import numpy as np

Zgate = np.array([[1, 0], [0, -1]])

zero = np.array([[1], [0]])
one = np.array([[0], [1]])

print(Zgate @ zero)
print(Zgate @ one)
```

Z|0> = |0>
Z|1> = -|1>

# Not Eigenstate

```python
from qiskit.visualization import plot_bloch_multivector

qc = QuantumCircuit(1)

qc.h(0)
qc.z(0)

state = Statevector.from_instruction(qc)

print(state)

plot_bloch_multivector(state)
```

H|0> = |+>

Z|+> = |->

Z|+> $\ne$ e^{i $\phi$}|+>

```python
import sympy as sp

phi = sp.symbols('phi',real=True)


p = sp.Matrix([[1, 0],
              [0, sp.exp(sp.I * phi)]])

eigenvalue = p.eigenvects()

print(eigenvalue)

```

## Key Takeaways 

- A gate can be apllied to any valid quantum gate.
- An eigenstate is a special state whose direction is preserved by the operation.
- The operation only changes its amplitude by an eigenvalue
- When the eigenvalue has the form e^{i $\phi$}, it represents a phase.
- This property is essential for phase kickback.
